**Practical-1: Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON files using Pandas.**

**Step 1: Import the Pandas Library**

```python
# Import the pandas library for data manipulation and analysis.
import pandas as pd
```

**Step 2: Create sample data**

```python
# Raw sample data containing student details (ID, Name, Age, City, Marks).

data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul","Priya","Amit","Riya","Karan","Krisha","Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat","Navsari","Valsad","Bharuch","Surat","Gandhinagar","Ahmedabad"],
    "Marks": [85, 92, 78, 88, 81, 84, 90]
}
```

**Step 3: Create a DataFrame**

```python
# Convert the dictionary into a Pandas DataFrame.
df = pd.DataFrame(data)
# Display the DataFrame in the output.
df
```

**Step 4: Save Data as a CSV file**

```python
# Save the DataFrame to a CSV file without the index column.
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

**Step 5: Save Data as an Excel file**

```python
# Save the DataFrame to a Excel file without the index column.
df.to_excel("students.xlsx", index=False)
print("Excel file create successfully.")
```

**Step 6: Save Data as a JSON file**

```python
# Save the DataFrame to a JSON file formatted as a list of record objects.
df.to_json("students.json",orient="records",indent=4)
print("JSON file created successfully.")
```

**Step 7: Read the CSV file**

```python
# Read data from the CSV file into a new DataFrame and display it.
df_csv = pd.read_csv("students.csv")
df_csv
```
