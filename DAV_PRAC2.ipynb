## **Practical-2: perform data inspection, cleaning, handling missing values duplicate removal, and data transformation using pandas.**

**Step 1: Import Required Library**

```python
# Import the pandas library.
import pandas as pd
```

**Step 2: Create a sample dataset**

```python
# Create a sample student dataset.
data = {
    "ID" : [101, 102, 103, 104, 104, 105, 106 ,107],
    "Name" : ["Rahul", "Priya", None ,"Amit", "Riya", "Ajay", "Naresh", "jenil"],
    "Age" : [20,21,None,20,20,22,23,28],
    "City" : ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahmedabad"],
    "Marks" : [85,None,78,88,88,81,90,93]
 }
```

```python
# Create a DataFrame from the dictionary.
df = pd.DataFrame(data)

# Display the DataFrame.
df
```

**Step 3: Inspect the Dataset**

```python
#Display the first 5 rows of the DataFrame.
df.head()
```

```python
# Display the last five rows of the dataset..
df.tail()
```

```python
# Display information of dataset.
df.info()
```

```python
# Display Statistical information of dataset.
df.describe()
```

```python
# Display the dimensions of dataset.
df.shape
```

```python
# Display the datatype of each column.
df.dtypes
```

```python
# Display the name of all column of dataset.
df.columns
```

**Step 4: check missing values**

```python
# check for missing (null) values in every column.
df.isnull()
```

```python
# Display the total number of missing values in each column.
df.isnull().sum()
```

```python
# print total null values for whole dataset.
df.isnull().sum().sum()
```

**Step 5: Fill Missing Values**

```python
# Calculate the average of the "Marks" column.
# and replace missing values with that average.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Display updated dataset.
df
```

```python
# Fill missing Age.
# Calculate the average of the "Age" column.
# And replace with that average.
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Display updated dataset.
df
```

```python
# Fill Missing Names.
# Calculate the most common name in the "Name" column.
# and replace missing values with that name.
df["Name"] = df["Name"].fillna("Unknown")

# Display updated Dataframe.
df
```

**Step 6: Detect duplicate records**

```python
#check wheter duplicate records.
df.duplicated()
```

```python
# Count the total number of duplicate rows.

df.duplicated().sum()
```

**Step 7: Remove duplicate records**

```python
# Remove duplicate rows from the dataset.
df = df.drop_duplicates()

# Display the updated dataset.
df
```

**Step 8: Rename a column**

```python
# Rename the "Marks" column to "Score".
df.rename(columns={"Marks":"Score"}, inplace=True)
# Display the updated dataset.
df
```

**Step 9: change data type**

```python
# Change the data type of the "Age" column to integer
df["Age"] = df["Age"].astype(int)

# Display the updated dataset
df
```

**Step 10: Create a New Column**

```python
# Create a new column "Final Score"
# Add 5 bonus marks to the "Score" column

df["Final Score"] = df["Score"] + 5

# Display the updated dataset
df
```

**Step 11: Convert Text To Uppercase**

```python
# Convert all names in the "Name" column to uppercase
df["Name"] = df["Name"].str.upper()

# Display the updated dataset
df
```

**Step 12: Convert text to lower**

```python
# Convert all city names in the "City" column to lowercase
df["City"] = df["City"].str.lower()
# Display the updated dataset
df
```

**Step - 13: Remove Extra Spaces**

```python
# Remove extra spaces from the "City" column
df["City"] = df["City"].str.strip()

# Display the updated dataset
df
```

**Step 14: Sort Dataset**

```python
# Sort the dataset by the "Age" column
df.sort_values(by="Age", inplace=True)
# Display the sorted dataset
df
```

**Step 15: Save the Cleaned Dataset**

```python
# Save the cleaned dataset
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
